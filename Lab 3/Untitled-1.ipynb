```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Raw data
# -----------------------------

# Orientation 1 data
distance_o1 = [5, 5.5, 6, 6.5, 7, 8, 8.6, 9]
successes_o1 = [30, 30, 30, 30, 30, 30, 12, 0]
success_rate_o1 = [100.0, 100.0, 100.0, 100.0, 100.0, 100.0, 40.0, 0.0]
avg_time_o1 = [35.7, 35.7, 35.7, 35.7, 35.7, 35.7, 36.0, np.nan]

# Fill these in after you collect Orientation 2 data
distance_o2 = []
successes_o2 = []
success_rate_o2 = []
avg_time_o2 = []

# Fill these in after you collect Orientation 3 data
distance_o3 = []
successes_o3 = []
success_rate_o3 = []
avg_time_o3 = []

# -----------------------------
# Create raw data table
# -----------------------------

data = []

for d, s, r, t in zip(distance_o1, successes_o1, success_rate_o1, avg_time_o1):
    data.append(["Orientation 1", d, s, r, t])

for d, s, r, t in zip(distance_o2, successes_o2, success_rate_o2, avg_time_o2):
    data.append(["Orientation 2", d, s, r, t])

for d, s, r, t in zip(distance_o3, successes_o3, success_rate_o3, avg_time_o3):
    data.append(["Orientation 3", d, s, r, t])

df = pd.DataFrame(data, columns=[
    "Orientation",
    "Distance (cm)",
    "Successful Reads",
    "Success Rate (%)",
    "Average Time to First Read (ms)"
])

print(df)

# Optional: save table to CSV
df.to_csv("coupling_study_raw_data.csv", index=False)

# -----------------------------
# Plot 1: Success rate vs distance
# -----------------------------

plt.figure(figsize=(8, 5))

plt.plot(distance_o1, success_rate_o1, marker="o", label="Orientation 1")

if len(distance_o2) > 0:
    plt.plot(distance_o2, success_rate_o2, marker="o", label="Orientation 2")

if len(distance_o3) > 0:
    plt.plot(distance_o3, success_rate_o3, marker="o", label="Orientation 3")

plt.xlabel("Distance (cm)")
plt.ylabel("Successful Reads (%)")
plt.title("Successful Reads vs Distance")
plt.ylim(-5, 105)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("success_rate_vs_distance.png", dpi=300)
plt.show()

# -----------------------------
# Plot 2: Average time to first read vs distance
# -----------------------------

plt.figure(figsize=(8, 5))

plt.plot(distance_o1, avg_time_o1, marker="o", label="Orientation 1")

if len(distance_o2) > 0:
    plt.plot(distance_o2, avg_time_o2, marker="o", label="Orientation 2")

if len(distance_o3) > 0:
    plt.plot(distance_o3, avg_time_o3, marker="o", label="Orientation 3")

plt.xlabel("Distance (cm)")
plt.ylabel("Average Time to First Read (ms)")
plt.title("Average Time to First Read vs Distance")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("avg_time_vs_distance.png", dpi=300)
plt.show()
```
